# Day 1 - End-to-End Churn Pipeline (RetailMind AI)

**Goal:** build, evaluate and save a churn model, then expose it through a `predict()` helper.

The reusable code lives in `src/` (loading, pipeline, prediction). This notebook is the **Day 1 record**: it runs each step and shows the results.

## 1. Setup
Make `src/` importable and load the reusable pieces.

```python
import sys
sys.path.append("..")

from src.data_loader import load_data, inspect, split_features_target
from src.pipeline import build_pipeline
from src.config import MODEL_PATH
```

## 2. Load and inspect the data

```python
df = load_data()
df.head()
```

```python
inspect(df)
```

**Findings**
- 7,043 customers, 21 columns.
- Only **26.5%** churn, so the classes are imbalanced and accuracy alone is misleading.
- `TotalCharges` was stored as text with **11 blank values** (new customers with tenure 0). `load_data()` converts it to numbers and the pipeline fills the blanks with the median.

```python
# Churn rate by contract type
df.groupby("Contract")["Churn"].apply(lambda s: (s == "Yes").mean()).round(3)
```

Month-to-month customers churn far more than customers on 1- or 2-year contracts.

## 3. Split into train and test

```python
from sklearn.model_selection import train_test_split

X, y = split_features_target(df)      # y: 1 = churned, 0 = stayed
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
print("Train rows:", len(X_train), "| Test rows:", len(X_test))
```

`stratify=y` keeps the churn ratio the same in both sets; `random_state=42` makes the split repeatable.

## 4. Build and train the pipeline
`build_pipeline()` (in `src/pipeline.py`) fills missing numbers, scales them, one-hot encodes the text columns, and ends with a Logistic Regression baseline.

```python
pipeline = build_pipeline(X_train)
pipeline
```

```python
pipeline.fit(X_train, y_train)
print("Model trained")
```

## 5. Evaluate on the test set

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

pred = pipeline.predict(X_test)
proba = pipeline.predict_proba(X_test)[:, 1]

for name, value in {
    "Accuracy": accuracy_score(y_test, pred),
    "Precision": precision_score(y_test, pred),
    "Recall": recall_score(y_test, pred),
    "F1": f1_score(y_test, pred),
    "ROC-AUC": roc_auc_score(y_test, proba),
}.items():
    print(f"{name}: {value:.4f}")
```

**Reading the numbers**
- Accuracy (~0.81) sounds good, but always guessing "No" would already give ~0.73.
- **Recall (~0.56)** is the weak spot: about 44% of customers who really leave are missed.
- **ROC-AUC (~0.84)** is the baseline every later day tries to beat.

## 6. Save the model

```python
import joblib

joblib.dump(pipeline, MODEL_PATH)
print("Saved to", MODEL_PATH.name)
```
